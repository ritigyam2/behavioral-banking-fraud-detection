## Evaluation Criteria

The dataset contains four imbalanced risk classes. Therefore, accuracy alone is not sufficient for selecting the final model.

The models are compared using Accuracy, Precision, Recall, F1-score and ROC-AUC. Macro, micro and weighted averages are included for multiclass evaluation.

**Macro F1 is used as the primary model-selection metric** because it gives equal importance to every risk class while balancing precision and recall.

## Model Evaluation Heatmap


```python
heat_metrics = [
    "Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1",
    "Weighted F1",
    "Macro ROC-AUC"
]

heat = results_df.set_index(
    "Model"
)[heat_metrics]

fig, ax = plt.subplots(
    figsize=(11, 4)
)

im = ax.imshow(
    heat.values,
    aspect="auto",
    vmin=0,
    vmax=1
)

ax.set_xticks(
    range(
        len(
            heat.columns
        )
    )
)

ax.set_xticklabels(
    heat.columns,
    rotation=40,
    ha="right"
)

ax.set_yticks(
    range(
        len(
            heat.index
        )
    )
)

ax.set_yticklabels(
    heat.index
)

for i in range(
    heat.shape[0]
):

    for j in range(
        heat.shape[1]
    ):

        ax.text(
            j,
            i,
            f"{heat.iloc[i, j]:.3f}",
            ha="center",
            va="center"
        )

plt.title(
    "Model Evaluation Heatmap"
)

plt.colorbar(im)

plt.tight_layout()
plt.show()
```

## Model Ranking


```python
ranking = results_df[
    [
        "Model",
        "Macro F1",
        "Macro Recall",
        "Macro ROC-AUC",
        "Accuracy",
        "Weighted F1"
    ]
].copy()

ranking["Macro F1 Rank"] = ranking[
    "Macro F1"
].rank(
    ascending=False,
    method="min"
)

ranking["Macro Recall Rank"] = ranking[
    "Macro Recall"
].rank(
    ascending=False,
    method="min"
)

ranking["ROC-AUC Rank"] = ranking[
    "Macro ROC-AUC"
].rank(
    ascending=False,
    method="min"
)

ranking = ranking.sort_values(
    [
        "Macro F1",
        "Macro Recall",
        "Macro ROC-AUC"
    ],
    ascending=False
)

display(
    ranking.round(4)
)
```

## Final Model Ranking


```python
final_ranking = results_df[
    [
        "Model",
        "Macro F1",
        "Macro Recall",
        "Accuracy",
        "Weighted F1",
        "Macro ROC-AUC"
    ]
].sort_values(
    [
        "Macro F1",
        "Macro Recall",
        "Macro ROC-AUC"
    ],
    ascending=False
).reset_index(
    drop=True
)

final_ranking.index = (
    final_ranking.index + 1
)

final_ranking.index.name = "Rank"

display(
    final_ranking.round(4)
)

best_model = (
    final_ranking.iloc[0]["Model"]
)

print(
    "Selected best model:",
    best_model
)
```

## Model-wise Interpretation

### Decision Tree

Decision Tree achieved **99.08% accuracy**, **51.46% Macro Precision**, **50.20% Macro Recall** and the highest **Macro F1-score of 50.13%**. Its Weighted F1-score was **98.65%**.

Although its Macro ROC-AUC of **83.50%** was lower than Random Forest and Logistic Regression, Decision Tree produced the strongest Macro F1 result.

### Random Forest

Random Forest achieved the highest **accuracy of 99.10%** and the highest **Macro ROC-AUC of 86.50%**. Its Macro F1-score was **49.68%**, which was only slightly lower than Decision Tree.

Random Forest is therefore a strong comparison model, particularly when overall accuracy and ROC-AUC are considered.

### Logistic Regression

Logistic Regression achieved **98.17% accuracy**, a Macro F1-score of **49.63%**, and the highest **Macro Recall of 51.11%**.

Its Macro ROC-AUC of **86.07%** was also strong. Logistic Regression therefore provides competitive and relatively balanced multiclass performance.

### Naive Bayes

Naive Bayes achieved **92.47% accuracy** and a Macro ROC-AUC of **85.91%**. However, its Macro Recall of **27.81%** and Macro F1-score of **28.78%** were considerably lower than the other models.

This indicates weaker balanced classification across the four risk levels despite relatively high overall accuracy.


## Conclusion and Final Model Selection

The four models were evaluated using Accuracy, Precision, Recall, F1-score and ROC-AUC. Since the dataset is highly imbalanced, accuracy alone was not considered sufficient for selecting the final model.

Greater importance was given to **Macro F1-score**, because it gives equal importance to Low, Moderate, High and Critical Risk while balancing precision and recall.

### Main Model: Decision Tree Classifier

**Decision Tree is selected as the main and final model.**

It achieved the highest Macro F1-score of **0.5013**, along with an accuracy of **99.08%** and Weighted F1-score of **98.65%**.

### Comparison Models

**Random Forest** is used as the first comparison model. It achieved the highest accuracy of **99.10%** and highest Macro ROC-AUC of **0.8650**, while its Macro F1-score was **0.4968**.

**Logistic Regression** is used as the second comparison model. It achieved the highest Macro Recall of **0.5111**, with a Macro F1-score of **0.4963** and Macro ROC-AUC of **0.8607**.

**Naive Bayes** is used as an additional baseline comparison model. Its Macro F1-score of **0.2878** and Macro Recall of **0.2781** indicate weaker balanced performance across the four classes.

### Final Decision

The leading three models are close and each has a different strength:

- **Decision Tree:** Highest Macro F1
- **Random Forest:** Highest Accuracy and Macro ROC-AUC
- **Logistic Regression:** Highest Macro Recall
- **Naive Bayes:** Baseline comparison model

Therefore, **Decision Tree Classifier is selected as the final model**, while Random Forest, Logistic Regression and Naive Bayes are retained as comparison models.

The selection is based primarily on balanced multiclass performance rather than accuracy alone, making it more appropriate for behavioral fraud-risk assessment.


## Limitations and Future Scope

The risk classes are highly imbalanced, with substantially more Low Risk transactions than High and Critical Risk transactions. This can produce very high overall accuracy even when performance on minority classes is weaker.

The final model should therefore be monitored using class-wise Precision, Recall and F1-score in addition to overall accuracy.

Future work can include stratified cross-validation, further class-weight tuning, probability calibration and evaluation on newer banking transactions as customer behaviour and fraud patterns change.
